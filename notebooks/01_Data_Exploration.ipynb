## 10. Target Variable Distribution

```python
plt.figure(figsize=(6,4))
sns.countplot(x="target", data=df)
plt.title("Distribution of Heart Disease")
plt.xlabel("Heart Disease")
plt.ylabel("Count")
plt.savefig("../reports/figures/target_distribution.png", dpi=300, bbox_inches="tight")
plt.show()
```

## 11. Age Distribution

```python
plt.figure(figsize=(8,5))
sns.histplot(df["age"], bins=20, kde=True)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.savefig("../reports/figures/age_distribution.png", dpi=300, bbox_inches="tight")
plt.show()
```

## 12. Gender Distribution

```python
plt.figure(figsize=(5,4))
sns.countplot(x="sex", data=df)
plt.xticks([0,1],["Female","Male"])
plt.title("Gender Distribution")
plt.savefig("../reports/figures/gender_distribution.png", dpi=300, bbox_inches="tight")
plt.show()
```

## 13. Chest Pain Types

```python
plt.figure(figsize=(7,4))
sns.countplot(x="cp", data=df)
plt.title("Chest Pain Types")
plt.savefig("../reports/figures/chest_pain_distribution.png", dpi=300, bbox_inches="tight")
plt.show()
```

## 14. Correlation Heatmap

```python
eda_df = df.replace("?", np.nan)
eda_df["ca"] = pd.to_numeric(eda_df["ca"])
eda_df["thal"] = pd.to_numeric(eda_df["thal"])

plt.figure(figsize=(12,8))
try:
    corr_matrix = eda_df.corr(numeric_only=True)
except TypeError:
    corr_matrix = eda_df.corr()

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Correlation Heatmap")
plt.savefig("../reports/figures/correlation_heatmap.png", dpi=300, bbox_inches="tight")
plt.show()
```
