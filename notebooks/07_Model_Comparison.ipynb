# Model Comparison

This notebook compares the performance of K-Nearest Neighbors, Logistic Regression, and Random Forest models on the Cleveland Heart Disease dataset.

## Step 127: Build a comparison table

```python
import pandas as pd
import matplotlib.pyplot as plt

comparison = pd.DataFrame({
    "Model": [
        "KNN",
        "Logistic Regression",
        "Random Forest"
    ],
    "Accuracy": [
        0.9016,
        0.8689,
        0.8852
    ],
    "Precision": [
        0.8235,
        0.8125,
        0.8182
    ],
    "Recall": [
        1.0000,
        0.9286,
        0.9643
    ],
    "Specificity": [
        0.8182,
        0.8182,
        0.8182
    ],
    "F1 Score": [
        0.9000,
        0.8667,
        0.8852
    ]
})

comparison
```

## Step 128: Accuracy Comparison

```python
plt.figure(figsize=(8,5))

plt.bar(
    comparison["Model"],
    comparison["Accuracy"],
    color=["blue", "green", "red"]
)

plt.title("Accuracy Comparison")

plt.ylabel("Accuracy")

plt.show()
```

## Step 129: Recall Comparison

```python
plt.figure(figsize=(8,5))

plt.bar(
    comparison["Model"],
    comparison["Recall"],
    color=["blue", "green", "red"]
)

plt.title("Recall Comparison")

plt.ylabel("Recall")

plt.show()
```

## Step 130: F1 Score Comparison

```python
plt.figure(figsize=(8,5))

plt.bar(
    comparison["Model"],
    comparison["F1 Score"],
    color=["blue", "green", "red"]
)

plt.title("F1 Score Comparison")

plt.ylabel("F1 Score")

plt.show()
```

## Model Selection

Although all three models achieved competitive performance, the selected model should be based on the evaluation metrics observed during experimentation.

For healthcare applications, **recall (sensitivity)** is particularly important because failing to identify a patient with heart disease (a false negative) may delay diagnosis and treatment.

Based on the experimental results, the model with the best balance of recall, precision, and overall performance would be the preferred choice for deployment.

Here, the **Tuned K-Nearest Neighbors (KNN with K=9)** model achieved the highest Accuracy (90.16%) and the highest F1-Score (90.00%) while maintaining a perfect Recall of 100.00%. Therefore, it is the selected model.
